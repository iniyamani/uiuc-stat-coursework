### 3. [3.5 points] Comparing Our Two Models for Overfitting

We now have two candidate models: our model from forward selection in Question 1 and our reduced model from Question 2.  In this question, we'll aim to identify which model performs better on *new* data.

**a)** Calculate the RMSE for each of our two models on each of our two data sets: the training and the test set.  Be sure that you use the same original model for both calculations (that is, you don't need to re-fit a second version of your model to your test data).

```python
full_model = smf.ols(formula="crew~cabins+length+passengers", data=df_train).fit()
full_model.rsquared_adj
```

```python
reduced_model = smf.ols(formula="crew~cabins+length", data=df_train).fit()
reduced_model.rsquared_adj
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

full_train_pred = full_model.predict(df_train)
full_test_pred = full_model.predict(df_test)

full_train_rmse = np.sqrt(mean_squared_error(df_train["crew"], full_train_pred))
full_test_rmse = np.sqrt(mean_squared_error(df_test["crew"], full_test_pred))

full_train_rmse, full_test_rmse
```

```python
full_train_pred = reduced_model.predict(df_train)
full_test_pred = reduced_model.predict(df_test)

full_train_rmse = np.sqrt(mean_squared_error(df_train["crew"], full_train_pred))
full_test_rmse = np.sqrt(mean_squared_error(df_test["crew"], full_test_pred))

full_train_rmse, full_test_rmse
```
